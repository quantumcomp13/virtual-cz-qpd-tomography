```python
import numpy as np
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit_aer.primitives import Estimator

def build_circuits_for_ZZ():


    circuits_zz = [[] for _ in range (6)]
    qr = QuantumRegister(2)
    cr_mcm = ClassicalRegister(1)
    cr = ClassicalRegister(2)

    for i in range (2):
        for j in range (2):
            for k in range (2):
                for l in range (2):
                    for c in range (6):
                        qc = QuantumCircuit(qr, cr_mcm, cr)
                        if i == 1:
                            qc.x(0)
                        if j == 1:
                            qc.h(0)
                        if k == 1:
                            qc.x(1)
                        if l == 1:
                            qc.h(1)
                        if c == 0:
                            qc.rz(np.pi/2, 0)
                            qc.rz(np.pi/2, 1)
                            qc.measure(0, cr[0])
                            qc.measure(1, cr[1])
                        if c == 1:
                            qc.rz(-np.pi/2, 0)
                            qc.rz(-np.pi/2, 1)
                            qc.measure(0, cr[0])
                            qc.measure(1, cr[1])
                        if c == 2:
                            qc.measure(1, cr_mcm[0])
                            qc.rz(-np.pi, 0)
                            qc.measure(0, cr[0])
                            qc.measure(1, cr[1])
                        if c == 3:
                            qc.measure(1, cr_mcm[0])
                            qc.rz(0, 0)
                            qc.measure(0, cr[0])
                            qc.measure(1, cr[1])
                        if c == 4:
                            qc.measure(0, cr_mcm[0])
                            qc.rz(-np.pi, 1)
                            qc.measure(0, cr[0])
                            qc.measure(1, cr[1])
                        if c == 5:
                            qc.measure(0, cr_mcm[0])
                            qc.rz(0, 1)
                            qc.measure(0, cr[0])
                            qc.measure(1, cr[1])
                        circuits_zz[c].append(qc)

    
    return circuits_zz
qc_zz = build_circuits_for_ZZ()
```

```python
def compute_expectation_from_counts(counts, has_mcm):
    shots = sum(counts.values()) #'counts' is for only 1 circuit not all the 6
    exp_array = np.zeros(3)
    for bitstring, count in counts.items(): #arranging bitstrings and corr counts
        prob = count / shots
        sep = bitstring.split(' ') #splitting the counts based on classical registers
        if has_mcm:
            bit_out = sep[0][::-1]
            bit_mcm = sep[1]
            mcm_val = 1 if bit_mcm == '0' else -1 #handling mcm
        else:
            bit_out = sep[0][::-1]
            mcm_val = 1
        z0 = 1 if bit_out[0] == '0' else -1
        z1 = 1 if bit_out[1] == '0' else -1
        exp_array[2] += z0 * z1 * prob * mcm_val
        exp_array[1] += z0 * prob * mcm_val
        exp_array[0] += z1 * prob * mcm_val
    return exp_array #this is a 1x3 array for each circuit
```

```python
from qiskit.quantum_info import Pauli


def execute_lo_circuits_zz(circuits, shots=5000):

    backend = AerSimulator()
    expectation_values= []
    counts_list = [[] for _ in range (16)] #16x6 matrix
    for i in range(len(circuits[0])):
        expectation_matrix = np.zeros((6, 3))
        for j in range (6):
            transpiled_circuits = transpile(circuits[j][i], backend) #transpiling jth circuit for ith input
            job = backend.run(transpiled_circuits, shots=shots) #job for jth circuit of ith input
            result = job.result() #result for jth circuit of ith input
            counts_zz = result.get_counts() #counts for jth circuit of ith input
            counts_list[i].append(counts_zz) #adding the counts jth circuit of ith input, 1x6 matrix
        for k in range (6): #till now all the 6 circuits of ith input have already ran
            if k<=1: #there is no mcm for first 2 circuits
                expectation_matrix[k] = compute_expectation_from_counts(counts_list[i][k], False)
                #passing counts of kth circuit of ith input
            elif k>=2 and k<=5: #there is mcm for rest of the 4 circuits
                expectation_matrix[k] = compute_expectation_from_counts(counts_list[i][k], True)
                #the function outputs 1x3 array for each kth circuit of ith input
                #for 6 circuits, it would 6x3 matrix for each input state
        expectation_values.append(expectation_matrix)
    return expectation_values, counts_list
#zz_exp is a 3d array of 16x6x3 dimensions. Its each row now has 6 circuits and each of them has a 1x3 array
#of expectation values. I have to extract expectation values of all the observables. First adjust 6x3 arrays
#with qpd coefficients, extract the expectation values for of all 16 observables for each input. Lastly build
#the 16x16 matrix where rows is the input state and the column is expectation value.

zz_expectation, counts_list_zz = execute_lo_circuits_zz(qc_zz)
```
